```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image

PROJECT_DIR = Path("..")
SPLITS_DIR = PROJECT_DIR / "data" / "splits"

train_df = pd.read_csv(SPLITS_DIR / "train_split.csv")
test_df = pd.read_csv(SPLITS_DIR / "test_split.csv")

print("Treino:", train_df.shape)
print("Teste:", test_df.shape)

train_df.head()
```

```python
print(train_df.columns.tolist())
```

```python
class_names = {
    0: "Sem retinopatia",
    1: "Retinopatia leve",
    2: "Retinopatia moderada",
    3: "Retinopatia severa",
    4: "Retinopatia proliferativa"
}
```

```python
sample = train_df.iloc[0]

image_path = Path(sample["image_path"])
label = sample["diagnosis"]

img = Image.open(image_path)

print("Caminho:", image_path)
print("Classe numérica:", label)
print("Classe:", class_names[label])
print("Tamanho da imagem:", img.size)

img
```

```python
samples = train_df.sample(9, random_state=42)

plt.figure(figsize=(10, 10))

for i, (_, row) in enumerate(samples.iterrows()):
    img = Image.open(row["image_path"])
    label = row["diagnosis"]

    plt.subplot(3, 3, i + 1)
    plt.imshow(img)
    plt.title(f"{label} - {class_names[label]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
